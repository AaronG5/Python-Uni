```python
import numpy as np 
import sympy as sp
from matplotlib import pyplot as plt

print(np.linspace(-1.3, 2.5, 64))
```

```python
def task_2():
   arr = np.array([1, 2, 3, 4])
   N = 3
   return np.tile(arr, N)

print(task_2())
```

<br>



```python
def task_3():
   arr = np.array([3, 4])
   N = 4
   return np.repeat(arr, N)

print(task_3())
```

<br>



```python
def task_4():
   arr = np.zeros((10, 10), dtype=int)
   return np.pad(arr, 1, constant_values=1)

print(task_4())
```

<br>



```python
def task_5():
   y, x = np.indices((8, 8), dtype=int)
   return (x + y) & 1

print(task_5())
```

<br>



```python
def task_6():
   n = 8
   i, j = np.indices((n, n), dtype=int)
   return i+j

print(task_6())
```

<br>



```python
def task_7():
   arr = np.random.rand(5, 5)

   return arr[arr[:, 1].argsort()]

print(task_7())
```

<br>



```python
def task_8():
   arr = np.random.randint(0, 3, (3, 3))
   try:
      res = np.linalg.eig(arr)
   except Exception:
      return None

   return res

print(task_8())
```

<br>



```python
def task_9():
   func = np.poly1d([0.5, 5, 4])
   deriv_1 = np.polyder(func, 1)
   deriv_2 = np.polyder(deriv_1, 1)

   return (deriv_1, deriv_2)

d1, d2 = task_9()
print(d1)
print(d2)
```

<br>



```python
def task_10():
   x = sp.symbols('x')
   f = sp.exp(-x)
   integral_indef = sp.integrate(f, x)
   integral_def = sp.integrate(f, (x, 0, 1))

   return (integral_indef, integral_def)

int_1, int_2 = task_10()
print(int_1)
print(int_2)
```

```python
def task_11():
   theta = np.linspace(0, 2*np.pi, 1000)
   a = 1
   r = 2 * a * (1 - np.cos(theta))

   plt.polar(theta, r)
   plt.show()

task_11()
```
